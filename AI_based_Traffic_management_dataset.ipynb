```python
import os
import shutil
from pathlib import Path
from collections import Counter, defaultdict

final_dataset_path = "/content/final_vehicle_traffic_dataset"

final_classes = [
    "Bike",
    "Car",
    "Auto",
    "Bus",
    "Truck",
    "Ambulance",
    "Emergency_Vehicle"
]

image_extensions = [".jpg", ".jpeg", ".png", ".bmp", ".webp"]

# Class image limits to reduce imbalance
# Emergency classes are not capped
class_image_limits = {
    "train": {
        0: 9000,   # Bike
        1: 9000,   # Car
        2: 7000,   # Auto
        3: 7000,   # Bus
        4: 7000,   # Truck
        5: None,   # Ambulance - keep all
        6: None    # Emergency_Vehicle - keep all
    },
    "valid": {
        0: 2000,
        1: 2000,
        2: 1500,
        3: 1500,
        4: 1500,
        5: None,
        6: None
    },
    "test": {
        0: 2000,
        1: 2000,
        2: 1500,
        3: 1500,
        4: 1500,
        5: None,
        6: None
    }
}

# Track how many images per class are already selected
selected_image_counts = {
    "train": Counter(),
    "valid": Counter(),
    "test": Counter()
}

def should_copy_image(target_split, mapped_class_ids):
    """
    Decide whether to copy an image based on class balancing limits.
    Emergency classes are always copied.
    Normal vehicle classes are capped.
    """

    mapped_class_ids = set(mapped_class_ids)

    # Always keep ambulance and emergency vehicle images
    if 5 in mapped_class_ids or 6 in mapped_class_ids:
        return True

    # For normal vehicles, copy only if at least one class is still below its limit
    for class_id in mapped_class_ids:
        limit = class_image_limits[target_split].get(class_id)

        if limit is None:
            return True

        if selected_image_counts[target_split][class_id] < limit:
            return True

    return False


def update_selected_counts(target_split, mapped_class_ids):
    """
    Update image-level class counts after copying an image.
    """
    for class_id in set(mapped_class_ids):
        selected_image_counts[target_split][class_id] += 1


def copy_yolo_dataset_with_balancing(
    source_root,
    source_split,
    target_split,
    class_map,
    prefix
):
    source_images_dir = os.path.join(source_root, source_split, "images")
    source_labels_dir = os.path.join(source_root, source_split, "labels")

    target_images_dir = os.path.join(final_dataset_path, target_split, "images")
    target_labels_dir = os.path.join(final_dataset_path, target_split, "labels")

    copied_images = 0
    copied_labels = 0
    skipped_images_due_to_balance = 0
    skipped_objects = 0
    missing_labels = 0
    empty_after_filter = 0
    invalid_lines = 0

    for image_file in os.listdir(source_images_dir):
        if Path(image_file).suffix.lower() not in image_extensions:
            continue

        image_stem = Path(image_file).stem
        label_file = image_stem + ".txt"

        source_image_path = os.path.join(source_images_dir, image_file)
        source_label_path = os.path.join(source_labels_dir, label_file)

        if not os.path.exists(source_label_path):
            missing_labels += 1
            continue

        new_lines = []
        mapped_class_ids = []

        with open(source_label_path, "r", errors="ignore") as f:
            lines = f.readlines()

        for line in lines:
            parts = line.strip().split()

            if len(parts) != 5:
                invalid_lines += 1
                continue

            try:
                old_class_id = int(float(parts[0]))
                x, y, w, h = map(float, parts[1:])
            except:
                invalid_lines += 1
                continue

            if old_class_id not in class_map:
                skipped_objects += 1
                continue

            if not (0 <= x <= 1 and 0 <= y <= 1 and 0 <= w <= 1 and 0 <= h <= 1):
                invalid_lines += 1
                continue

            new_class_id = class_map[old_class_id]
            mapped_class_ids.append(new_class_id)
            new_lines.append(f"{new_class_id} {x} {y} {w} {h}")

        if len(new_lines) == 0:
            empty_after_filter += 1
            continue

        if not should_copy_image(target_split, mapped_class_ids):
            skipped_images_due_to_balance += 1
            continue

        new_image_name = f"{prefix}_{image_file}"
        new_label_name = f"{prefix}_{label_file}"

        target_image_path = os.path.join(target_images_dir, new_image_name)
        target_label_path = os.path.join(target_labels_dir, new_label_name)

        shutil.copy2(source_image_path, target_image_path)

        with open(target_label_path, "w") as f:
            f.write("\n".join(new_lines))

        update_selected_counts(target_split, mapped_class_ids)

        copied_images += 1
        copied_labels += 1

    return {
        "copied_images": copied_images,
        "copied_labels": copied_labels,
        "skipped_images_due_to_balance": skipped_images_due_to_balance,
        "skipped_objects": skipped_objects,
        "missing_labels": missing_labels,
        "empty_after_filter": empty_after_filter,
        "invalid_lines": invalid_lines
    }

print("Balanced merge function created successfully.")
print("Final classes:")
for idx, cls in enumerate(final_classes):
    print(idx, ":", cls)
```

```python
normal_root = "/content/raw_datasets/normal_vehicles/vehicle dataset"

# Source class ID -> Final class ID
normal_class_map = {
    0: 1,  # car -> Car
    1: 2,  # threewheel -> Auto
    2: 3,  # bus -> Bus
    3: 4,  # truck -> Truck
    4: 0,  # motorbike -> Bike
    5: 1   # van -> Car
}

results = {}

results["train"] = copy_yolo_dataset_with_balancing(
    source_root=normal_root,
    source_split="train",
    target_split="train",
    class_map=normal_class_map,
    prefix="normal"
)

results["valid"] = copy_yolo_dataset_with_balancing(
    source_root=normal_root,
    source_split="valid",
    target_split="valid",
    class_map=normal_class_map,
    prefix="normal"
)

print("Normal vehicle dataset merged with balancing.")

for split, result in results.items():
    print(f"\n{split.upper()}:")
    for key, value in result.items():
        print(key + ":", value)

print("\nCurrent selected image counts:")
for split in ["train", "valid", "test"]:
    print("\n" + split.upper())
    for class_id, class_name in enumerate(final_classes):
        print(f"{class_id} - {class_name}: {selected_image_counts[split][class_id]}")
```

```python
import os
from pathlib import Path

ev_nev_root = "/content/raw_datasets/ev_nev_annotated/final_split2"
image_extensions = [".jpg", ".jpeg", ".png", ".bmp", ".webp"]

removed_labels = 0
removed_images = 0
invalid_files = []

for split in ["train", "val"]:
    labels_dir = os.path.join(ev_nev_root, split, "labels")
    images_dir = os.path.join(ev_nev_root, split, "images")

    for label_file in os.listdir(labels_dir):
        if not label_file.endswith(".txt"):
            continue

        label_path = os.path.join(labels_dir, label_file)
        is_invalid = False

        with open(label_path, "r", errors="ignore") as f:
            lines = f.readlines()

        for line in lines:
            parts = line.strip().split()

            if len(parts) == 0:
                continue

            if len(parts) != 5:
                is_invalid = True
                break

            try:
                int(float(parts[0]))
                float(parts[1])
                float(parts[2])
                float(parts[3])
                float(parts[4])
            except:
                is_invalid = True
                break

        if is_invalid:
            invalid_files.append((split, label_file))
            os.remove(label_path)
            removed_labels += 1

            image_stem = Path(label_file).stem
            for ext in image_extensions:
                image_path = os.path.join(images_dir, image_stem + ext)
                if os.path.exists(image_path):
                    os.remove(image_path)
                    removed_images += 1
                    break

print("Invalid EV label cleanup completed.")
print("Invalid files found:", len(invalid_files))
print("Removed label files:", removed_labels)
print("Removed matching images:", removed_images)

if invalid_files:
    print("\nRemoved files:")
    for split, label_file in invalid_files:
        print(split, "->", label_file)
```

```python
import os
from collections import Counter

final_dataset_path = "/content/final_vehicle_traffic_dataset"

final_classes = [
    "Bike",
    "Car",
    "Auto",
    "Bus",
    "Truck",
    "Ambulance",
    "Emergency_Vehicle"
]

class_counter = Counter()
invalid_label_files = []
invalid_lines = []
total_objects = 0

for split in ["train", "valid", "test"]:
    labels_dir = os.path.join(final_dataset_path, split, "labels")

    for label_file in os.listdir(labels_dir):
        if not label_file.endswith(".txt"):
            continue

        label_path = os.path.join(labels_dir, label_file)

        with open(label_path, "r", errors="ignore") as f:
            lines = f.readlines()

        for line_number, line in enumerate(lines, start=1):
            parts = line.strip().split()

            if len(parts) != 5:
                invalid_label_files.append(label_path)
                invalid_lines.append((split, label_file, line_number, line.strip()))
                continue

            try:
                class_id = int(float(parts[0]))
                x, y, w, h = map(float, parts[1:])
            except:
                invalid_label_files.append(label_path)
                invalid_lines.append((split, label_file, line_number, line.strip()))
                continue

            if class_id < 0 or class_id >= len(final_classes):
                invalid_label_files.append(label_path)
                invalid_lines.append((split, label_file, line_number, line.strip()))
                continue

            if not (0 <= x <= 1 and 0 <= y <= 1 and 0 <= w <= 1 and 0 <= h <= 1):
                invalid_label_files.append(label_path)
                invalid_lines.append((split, label_file, line_number, line.strip()))
                continue

            class_counter[class_id] += 1
            total_objects += 1

print("Final dataset label validation completed.")
print("Total objects:", total_objects)
print("Invalid label files:", len(set(invalid_label_files)))
print("Invalid label lines:", len(invalid_lines))

print("\nClass distribution:")
for class_id, class_name in enumerate(final_classes):
    print(f"{class_id} - {class_name}: {class_counter[class_id]} objects")

if invalid_lines:
    print("\nFirst 20 invalid lines:")
    for item in invalid_lines[:20]:
        split, label_file, line_number, text = item
        print(f"{split} | {label_file} | line {line_number}: {text}")
```
