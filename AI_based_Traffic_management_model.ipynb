```python
import zipfile
import os
import shutil

local_zip_path = "/content/final_vehicle_traffic_dataset.zip"
extract_path = "/content"

dataset_path = "/content/final_vehicle_traffic_dataset"

# Remove old extracted dataset if it exists
if os.path.exists(dataset_path):
    shutil.rmtree(dataset_path)
    print("Old extracted dataset removed.")

# Extract ZIP
with zipfile.ZipFile(local_zip_path, "r") as zip_ref:
    zip_ref.extractall(extract_path)

print("Dataset extracted successfully.")
print("Dataset path:", dataset_path)
print("Dataset exists:", os.path.exists(dataset_path))
print("data.yaml exists:", os.path.exists(os.path.join(dataset_path, "data.yaml")))

print("\nMain items inside dataset folder:")
for item in os.listdir(dataset_path):
    print(item)
```

```python
import os

dataset_path = "/content/final_vehicle_traffic_dataset"

for split in ["train", "valid", "test"]:
    images_dir = os.path.join(dataset_path, split, "images")
    labels_dir = os.path.join(dataset_path, split, "labels")

    image_files = [
        f for f in os.listdir(images_dir)
        if f.lower().endswith((".jpg", ".jpeg", ".png", ".bmp", ".webp"))
    ]

    label_files = [
        f for f in os.listdir(labels_dir)
        if f.lower().endswith(".txt")
    ]

    print(f"{split.upper()}:")
    print("Images:", len(image_files))
    print("Labels:", len(label_files))
    print("Match:", len(image_files) == len(label_files))
    print("-" * 30)

print("data.yaml path:")
print(os.path.join(dataset_path, "data.yaml"))
```

```python
from ultralytics import YOLO

data_yaml_path = "/content/final_vehicle_traffic_dataset/data.yaml"

model = YOLO("yolov8s.pt")

results = model.train(
    data=data_yaml_path,
    epochs=4,
    imgsz=640,
    batch=32,
    device=0,
    workers=4,
    cache=True,
    patience=8,
    project="/content/traffic_training_runs",
    name="yolov8s_vehicle_emergency_fast",
    exist_ok=True
)

print("Training completed.")
```

```python
from ultralytics import YOLO

model = YOLO("/content/traffic_training_runs/yolov8s_vehicle_emergency_fast/weights/last.pt")

results = model.train(
    data="/content/final_vehicle_traffic_dataset/data.yaml",
    epochs=5,
    imgsz=640,
    batch=32,
    device=0,
    workers=4,
    cache=True,
    patience=5,
    project="/content/traffic_training_runs",
    name="yolov8s_vehicle_emergency_chunk_2",
    exist_ok=True
)

print("5-epoch chunk completed.")
```
