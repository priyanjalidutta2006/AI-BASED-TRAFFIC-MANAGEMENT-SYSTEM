```python

import os

dataset_path = "/content/kaggle_dataset_download/final_vehicle_traffic_dataset"
data_yaml_path = os.path.join(dataset_path, "data.yaml")

print("Dataset path:", dataset_path)
print("Dataset path exists:", os.path.exists(dataset_path))
print("data.yaml exists:", os.path.exists(data_yaml_path))

print("\nMain items inside dataset folder:")
if os.path.exists(dataset_path):
    for item in os.listdir(dataset_path):
        item_path = os.path.join(dataset_path, item)
        if os.path.isdir(item_path):
            print("[FOLDER]", item)
        else:
            print("[FILE]", item)

print("\nSplit-wise image and label counts:")

for split in ["train", "valid", "test"]:
    images_dir = os.path.join(dataset_path, split, "images")
    labels_dir = os.path.join(dataset_path, split, "labels")

    print(f"\n{split.upper()}:")
    print("Images folder exists:", os.path.exists(images_dir))
    print("Labels folder exists:", os.path.exists(labels_dir))

    if os.path.exists(images_dir):
        image_count = len([
            f for f in os.listdir(images_dir)
            if f.lower().endswith((".jpg", ".jpeg", ".png", ".bmp", ".webp"))
        ])
        print("Images:", image_count)

    if os.path.exists(labels_dir):
        label_count = len([
            f for f in os.listdir(labels_dir)
            if f.lower().endswith(".txt")
        ])
        print("Labels:", label_count)

print("\nCurrent data.yaml content:")
if os.path.exists(data_yaml_path):
    with open(data_yaml_path, "r") as f:
        print(f.read())
else:
    print("data.yaml not found.")
```

```python
from ultralytics import YOLO

data_yaml_path = "/content/kaggle_dataset_download/final_vehicle_traffic_dataset/data.yaml"
last_model_path = "/content/drive/MyDrive/AI BASED TRAFFIC MANAGEMENT SYSTEM/models/last.pt"

model = YOLO(last_model_path)

results = model.train(
    data=data_yaml_path,
    epochs=5,
    imgsz=640,
    batch=32,
    device=0,
    workers=4,
    cache=False,
    patience=5,
    project="/content/traffic_training_runs",
    name="yolov8s_vehicle_emergency_retrain_chunk_2",
    exist_ok=True
)

print("5-epoch retraining completed.")
```

```python
import os

runs_dir = "/content/traffic_training_runs"

for root, dirs, files in os.walk(runs_dir):
    if "best.pt" in files or "last.pt" in files:
        print("Found weights folder:", root)
        print("Files:", files)
        print("-" * 50)
```

```python
import os
import shutil

run_weights = "/content/traffic_training_runs/yolov8s_vehicle_emergency_retrain_chunk_2/weights"
drive_models = "/content/drive/MyDrive/AI BASED TRAFFIC MANAGEMENT SYSTEM/models"

print("Run weights folder exists:", os.path.exists(run_weights))
print("Files inside weights folder:", os.listdir(run_weights) if os.path.exists(run_weights) else "Not found")

os.makedirs(drive_models, exist_ok=True)

shutil.copy2(os.path.join(run_weights, "best.pt"), os.path.join(drive_models, "best.pt"))
shutil.copy2(os.path.join(run_weights, "last.pt"), os.path.join(drive_models, "last.pt"))

print("Drive model files updated successfully.")
print("Updated best.pt:", os.path.join(drive_models, "best.pt"))
print("Updated last.pt:", os.path.join(drive_models, "last.pt"))
```
